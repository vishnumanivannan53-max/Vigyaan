```python
from PIL import Image

image_path = "/kaggle/input/competitions/understanding_cloud_organization/train_images/397c2fd.jpg"

image = Image.open(image_path).convert("RGB")

print("Image loaded:", image.size)
```

```python
import torch
from transformers import BlipProcessor, BlipForQuestionAnswering

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Using:", device)

processor = BlipProcessor.from_pretrained(
    "Salesforce/blip-vqa-base"
)

model = BlipForQuestionAnswering.from_pretrained(
    "Salesforce/blip-vqa-base"
).to(device)

print("BLIP-VQA ready!")
```

```python
question = "What is visible in this image?"

inputs = processor(
    image,
    question,
    return_tensors="pt"
).to(device)

with torch.no_grad():
    output = model.generate(
        **inputs,
        max_new_tokens=20
    )

answer = processor.decode(
    output[0],
    skip_special_tokens=True
)

print("Question:", question)
print("Answer:", answer)
```

```python
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import os
import torch

image_dir = "/kaggle/input/datasets/alienxc137/earthvqa-semantic-segmentation-visual-question-ans/Train-003/Train/images_png"


class EarthVQADataset(Dataset):
    def __init__(self, records, processor, image_dir):
        self.records = records
        self.processor = processor
        self.image_dir = image_dir

    def __len__(self):
        return len(self.records)

    def __getitem__(self, idx):
        record = self.records[idx]

        image_path = os.path.join(
            self.image_dir,
            record["image"]
        )

        image = Image.open(image_path).convert("RGB")

        question = record["question"]
        answer = record["answer"]

        encoding = self.processor(
            images=image,
            text=question,
            return_tensors="pt",
            padding="max_length",
            truncation=True,
            max_length=32
        )

        labels = self.processor.tokenizer(
            answer,
            return_tensors="pt",
            padding="max_length",
            truncation=True,
            max_length=16
        ).input_ids

        # Ignore padding tokens when calculating loss
        labels[labels == self.processor.tokenizer.pad_token_id] = -100

        encoding = {k: v.squeeze(0) for k, v in encoding.items()}
        encoding["labels"] = labels.squeeze(0)

        return encoding
```

```python
import os
import re
import string
import torch
from PIL import Image
from collections import defaultdict


def normalize_text(s):
    s = str(s).lower().strip()
    s = re.sub(f"[{re.escape(string.punctuation)}]", "", s)
    s = re.sub(r"\s+", " ", s)
    return s.strip()


def classify_question_type(qa_type, question, answer):
    a = str(answer).lower().strip()
    t = str(qa_type).lower()

    if a in {"yes", "no"}:
        return "yes_no"
    if "rural" in a or "urban" in a:
        return "rural_urban"
    if "%" in a or re.fullmatch(r"[\d.]+", a):
        return "numeric_percentage"
    if len(a.split()) > 4 or "reasoning" in t or "situation" in t:
        return "descriptive_reasoning"
    return "land_use_category"


def generate_answer(model, processor, image, question, device,
                     max_new_tokens=20, num_beams=5,
                     repetition_penalty=1.5, no_repeat_ngram_size=3):
    inputs = processor(images=image, text=str(question), return_tensors="pt").to(device)
    with torch.no_grad():
        out = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            num_beams=num_beams,
            repetition_penalty=repetition_penalty,
            no_repeat_ngram_size=no_repeat_ngram_size,
            early_stopping=True,
        )
    return processor.decode(out[0], skip_special_tokens=True).strip()


def evaluate_model(model, processor, records_subset, image_dir, device,
                    label="model", max_samples=None, verbose_examples=5):
    model.eval()
    records_to_use = records_subset if max_samples is None else records_subset[:max_samples]

    exact_correct = 0
    normalized_correct = 0
    total = 0

    per_type_exact = defaultdict(int)
    per_type_normalized = defaultdict(int)
    per_type_total = defaultdict(int)

    correct_examples = []
    incorrect_examples = []

    for item in records_to_use:
        image_path = os.path.join(image_dir, item["image"])
        try:
            image = Image.open(image_path).convert("RGB")
        except Exception:
            continue

        question = str(item["question"])
        expected = str(item["answer"])
        qtype = classify_question_type(item.get("type", ""), question, expected)

        predicted = generate_answer(model, processor, image, question, device)

        exact = predicted.strip().lower() == expected.strip().lower()
        norm_match = normalize_text(predicted) == normalize_text(expected)

        total += 1
        per_type_total[qtype] += 1
        if exact:
            exact_correct += 1
            per_type_exact[qtype] += 1
        if norm_match:
            normalized_correct += 1
            per_type_normalized[qtype] += 1

        record = {
            "image": item["image"], "question": question,
            "expected": expected, "predicted": predicted, "type": qtype
        }
        (correct_examples if norm_match else incorrect_examples).append(record)

    results = {
        "label": label,
        "total": total,
        "exact_match_acc": exact_correct / total if total else 0,
        "normalized_acc": normalized_correct / total if total else 0,
        "per_type": {
            qt: {
                "count": per_type_total[qt],
                "exact_match_acc": per_type_exact[qt] / per_type_total[qt],
                "normalized_acc": per_type_normalized[qt] / per_type_total[qt],
            }
            for qt in per_type_total
        },
        "correct_examples": correct_examples[:verbose_examples],
        "incorrect_examples": incorrect_examples[:verbose_examples],
    }

    print(f"\n=== Evaluation: {label} ===")
    print(f"Total questions: {total}")
    print(f"Exact-match accuracy:  {results['exact_match_acc']*100:.2f}%")
    print(f"Normalized accuracy:   {results['normalized_acc']*100:.2f}%")
    print("\nPer-question-type breakdown:")
    for qt, stats in results["per_type"].items():
        print(f"  {qt:22s} n={stats['count']:4d}  exact={stats['exact_match_acc']*100:5.1f}%  "
              f"normalized={stats['normalized_acc']*100:5.1f}%")

    return results


print("Evaluation utilities ready: normalize_text, classify_question_type, generate_answer, evaluate_model")
```

```python
import os
import re
import string
import torch
from PIL import Image
from collections import defaultdict


def normalize_text(s):
    s = str(s).lower().strip()
    s = re.sub(f"[{re.escape(string.punctuation)}]", "", s)
    s = re.sub(r"\s+", " ", s)
    return s.strip()


def classify_question_type(qa_type, question, answer):
    a = str(answer).lower().strip()
    t = str(qa_type).lower()

    if a in {"yes", "no"}:
        return "yes_no"
    if "rural" in a or "urban" in a:
        return "rural_urban"
    if "%" in a or re.fullmatch(r"[\d.]+", a):
        return "numeric_percentage"
    if len(a.split()) > 4 or "reasoning" in t or "situation" in t:
        return "descriptive_reasoning"
    return "land_use_category"


def generate_answer(model, processor, image, question, device,
                     max_new_tokens=20, num_beams=5,
                     repetition_penalty=1.5, no_repeat_ngram_size=3):
    inputs = processor(images=image, text=str(question), return_tensors="pt").to(device)
    with torch.no_grad():
        out = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            num_beams=num_beams,
            repetition_penalty=repetition_penalty,
            no_repeat_ngram_size=no_repeat_ngram_size,
            early_stopping=True,
        )
    return processor.decode(out[0], skip_special_tokens=True).strip()


def evaluate_model(model, processor, records_subset, image_dir, device,
                    label="model", max_samples=None, verbose_examples=5):
    model.eval()
    records_to_use = records_subset if max_samples is None else records_subset[:max_samples]

    exact_correct = 0
    normalized_correct = 0
    total = 0

    per_type_exact = defaultdict(int)
    per_type_normalized = defaultdict(int)
    per_type_total = defaultdict(int)

    correct_examples = []
    incorrect_examples = []

    for item in records_to_use:
        image_path = os.path.join(image_dir, item["image"])
        try:
            image = Image.open(image_path).convert("RGB")
        except Exception:
            continue

        question = str(item["question"])
        expected = str(item["answer"])
        qtype = classify_question_type(item.get("type", ""), question, expected)

        predicted = generate_answer(model, processor, image, question, device)

        exact = predicted.strip().lower() == expected.strip().lower()
        norm_match = normalize_text(predicted) == normalize_text(expected)

        total += 1
        per_type_total[qtype] += 1
        if exact:
            exact_correct += 1
            per_type_exact[qtype] += 1
        if norm_match:
            normalized_correct += 1
            per_type_normalized[qtype] += 1

        record = {
            "image": item["image"], "question": question,
            "expected": expected, "predicted": predicted, "type": qtype
        }
        (correct_examples if norm_match else incorrect_examples).append(record)

    results = {
        "label": label,
        "total": total,
        "exact_match_acc": exact_correct / total if total else 0,
        "normalized_acc": normalized_correct / total if total else 0,
        "per_type": {
            qt: {
                "count": per_type_total[qt],
                "exact_match_acc": per_type_exact[qt] / per_type_total[qt],
                "normalized_acc": per_type_normalized[qt] / per_type_total[qt],
            }
            for qt in per_type_total
        },
        "correct_examples": correct_examples[:verbose_examples],
        "incorrect_examples": incorrect_examples[:verbose_examples],
    }

    print(f"\n=== Evaluation: {label} ===")
    print(f"Total questions: {total}")
    print(f"Exact-match accuracy:  {results['exact_match_acc']*100:.2f}%")
    print(f"Normalized accuracy:   {results['normalized_acc']*100:.2f}%")
    print("\nPer-question-type breakdown:")
    for qt, stats in results["per_type"].items():
        print(f"  {qt:22s} n={stats['count']:4d}  exact={stats['exact_match_acc']*100:5.1f}%  "
              f"normalized={stats['normalized_acc']*100:5.1f}%")

    return results


print("Evaluation utilities re-defined.")
```

```python
def classify_question_type(qa_type, question, answer):
    """
    Classifies EarthVQA questions using the question, answer,
    and original dataset type.
    """

    answer_text = str(answer).lower().strip()
    question_text = str(question).lower().strip()
    type_text = str(qa_type).lower().strip()

    # Yes/no answers
    if answer_text in {"yes", "no"}:
        return "yes_no"

    # Explicit rural/urban classification questions
    if (
        ("rural" in question_text and "urban" in question_text)
        or question_text.startswith("is it a rural")
        or answer_text in {"rural", "urban"}
    ):
        return "rural_urban"

    # Percentage or numerical questions
    if (
        "%" in answer_text
        or re.fullmatch(r"\d+(\.\d+)?", answer_text)
        or "area" in question_text
        or "percentage" in question_text
        or "percent" in question_text
        or "how many" in question_text
        or "number of" in question_text
        or "count" in question_text
    ):
        return "numeric_percentage"

    # Descriptive and reasoning questions
    if (
        len(answer_text.split()) > 4
        or "reasoning" in type_text
        or "situation" in type_text
        or "what are the needs" in question_text
        or "what are the land use types" in question_text
        or "what are the road materials" in question_text
        or "what are the water situations" in question_text
        or "describe" in question_text
        or "explain" in question_text
        or "why" in question_text
    ):
        return "descriptive_reasoning"

    return "land_use_category"
```

```python
class EarthVQADataset(Dataset):
    def __init__(self, records, image_dir, processor):
        self.records = records
        self.image_dir = image_dir
        self.processor = processor

    def __len__(self):
        return len(self.records)

    def __getitem__(self, idx):
        item = self.records[idx]

        image_path = os.path.join(
            self.image_dir,
            str(item["image"])
        )

        if not os.path.exists(image_path):
            raise FileNotFoundError(
                f"Image not found: {image_path}"
            )

        image = Image.open(image_path).convert("RGB")

        # Process image and question
        encoding = self.processor(
            images=image,
            text=str(item["question"]),
            return_tensors="pt",
            padding="max_length",
            truncation=True,
            max_length=16
        )

        # Process answer as training labels
        labels = self.processor.tokenizer(
            str(item["answer"]),
            return_tensors="pt",
            padding="max_length",
            truncation=True,
            max_length=16
        ).input_ids

        # Ignore padding tokens in the loss calculation
        labels[
            labels == self.processor.tokenizer.pad_token_id
        ] = -100

        return {
            "pixel_values": encoding["pixel_values"].squeeze(0),
            "input_ids": encoding["input_ids"].squeeze(0),
            "attention_mask": encoding["attention_mask"].squeeze(0),
            "labels": labels.squeeze(0)
        }
```

```python
import torch
from PIL import Image

def ask_question(image_path, question, show_image=False):
    image = Image.open(image_path).convert("RGB")

    inputs = processor(
        images=image,
        text=question,
        return_tensors="pt"
    )

    inputs = {
        key: value.to(DEVICE)
        for key, value in inputs.items()
    }

    with torch.no_grad():
        output_ids = model.generate(
            **inputs,
            max_new_tokens=20
        )

    predicted_answer = processor.decode(
        output_ids[0],
        skip_special_tokens=True
    ).strip()

    if show_image:
        display(image)

    print("Question:", question)
    print("Predicted answer:", predicted_answer)

    return predicted_answer
```
